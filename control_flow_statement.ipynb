# if-elif-else condition

```python
""" enter two values from user and if the value will be greater print greater, if the value is not greater print no greater, if equal print value is equal """
```

```python
num1 = int(input("Enter first value"))
num2 = int(input("Enter second value"))
if num1 > num2:
    print("Value first is greater, and that is", num1)
elif num1 < num2:
    print("Value second is greater, and that is", num2)
else:
    print("Value first and second are equal, and that is", num1, num2)
```

```python
""" take a value from user, and check it is divisible by 2 or 3 neither 2 or 3 """
```

```python
num = int(input("Enter a value"))
if num%2==0:
    print("Number is divisible by 2")
elif num%3==0:
    print("Number is divisible by 3")
else:
    print("Number is not divisible by 2 and 3")
```

```python
""" enter the marks of student, if marks are greater than or euqal to 85 print A grade, if marks are greater than or euqal to 75 print B grade, if marks are greater than or euqal to 65 print C grade, if marks are greater than or euqal to 55 print D grade, if marks are greater than or euqal to 45 print E grade """
```

```python
marks = int(input("Enter student marks"))
if marks >= 85:
    print("A Grade")
elif marks >=75 and marks < 85:
    print("B Grade")
elif marks >=65 and marks < 75:
    print("C Grade")
elif marks >=55 and marks < 65:
    print("D Grade")
elif marks >=45 and marks < 55:
    print("E Grade")
else:
    print("Fail")
```

# nested if condition

```python
""" enter the value from user, check if it is greater than 10 or 20 and not """
```

```python
value = int(input("Enter a value"))
if value >=10:
    if value >=20:
        print("Value is greater than 10 and 20 and value is", value)
    else:
        print("Value is less than 20")
else:
    print("Value is not greater than 10 and 20")
```
