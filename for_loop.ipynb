# for loop in python

```python
""" Write a  program to printthe square of the number till 10 """
```

```python
L = []
for i in range(10):
    print(i+1) # starts with 0
    L.append(i**2)
print(L)
```

```python
L = []
for i in range(0,10,2):
    print(i+1) # starts with 0
    L.append(i**2)
print(L)
```

```python
""" Write a program to separate positive and negative values"""
```

```python
L = [24, 15, 67, 59, -2, -12, -34, 89, 78, 44]
L1 = [] # positive numbers stores 
L2 = [] # negative numbers stores
for i in range(10):
    if L[i]>=0:
        L1.append(L[i])
    else:
        L2.append(L[i])
print("The list of positive values are:", L1)
print("The list of negative values are:", L2)
```

```python
""" Write a program to separate the even and odd numbers in list"""
```

```python
L = [40, 52, 45, 91, 56, 23, 43, 89, 88, 44]
L1 = [] # even numbers stores 
L2 = [] # odd numbers stores
for i in range(10):
    if L[i]%2==0:
        L1.append(L[i])
    else:
        L2.append(L[i])
print("The list of even values are:", L1)
print("The list of odd values are:", L2)
```

```python
# for with else 
S =  {"apple", "bannana", 2.56, "pen", "pencil", 4}
for x in S:
    print(x)
else:
    print("loop terminated")
print("done")
```

```python
D = {"A":34, "B":56, 44 : "C"}
for x  in D:
    print(x, D[x])
```

# Nesting of for loop

```python
""" ask user to enter a string and filter count vowels in the string"""
```

```python
L = []
s = input("Enter a statement: ")
for i in range(len(s)):
    L.append(s[i])
print(L)
count = 0
for i in range(len(L)):
    if L[i] == "a":
        print("a vowel is present")
        count=count+1
        print(count)
for i in range(len(L)):
    if L[i] == "e":
        print("e vowel is present")
        count=count+1
        print(count)
for i in range(len(L)):
    if L[i] == "i":
        print("i vowel is present")
        count=count+1
        print(count)
for i in range(len(L)):
    if L[i] == "o":
        print("o vowel is present")
        count=count+1
        print(count)
for i in range(len(L)):
    if L[i] == "u":
        print("u vowel is present")
        count=count+1
        print(count)
```
