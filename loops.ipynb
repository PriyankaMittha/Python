# while loop

```python
""" take the number from the user and print all numbers in between 1 to that """
```

```python
n = int(input("Enter a number:"))
i = 1
while i <= n :
    print(i)
    i = i + 1
print("The loop work is done")
```

```python
""" Enter the value and write its multiplication table till 10 """
```

```python
n = int(input("Enter a number:"))
i = 1
while i <= 10:
    print(n*i)
    i = i+1
print("Table Done")
```

```python
""" Enter a number and print its exponential value till power 5 """
```

```python
n = int(input("Enter a value:"))
i = 1
while i <= 5:
    print(n**i)
    i = i+1
print("Exponential done")
```

```python
""" Enter a number, find out the even numbers in between that """
```

```python
n = int(input("Enter a value:"))
i = 1
while i <= n:
    if i % 2 == 0:
        print(i)
    else:
        pass
    i = i+1
print("Done")
```

```python
""" print a pattern of *** """
```

```python
n = int(input("Enter the no.of rows:"))
i = 1
while i <= n:
    j = 1
    while j <= i:
        print("*", end = " ")
        j = j+1
    print()
    i = i+1
```

```python
""" pattern in reverse """
```

```python
n = int(input("Enter the no.of rows:"))
i = 1
while i <= n:
    j = n
    while j >= i:
        print("*", end = " ")
        j = j-1
    print()
    i = i+1
```

# Break Statement

```python
n = 10
i = 1
while True:
    if i % 9 == 0:
        print("Inside if")
        break
    else:
        print("Inside else")
        i = i+1
print("Done")
```

```python
""" print the value till 10 """
```

```python
n = 10
i = 1
while i <= n:
    print(i)
    i = i+1
    if i == 11:
        break
print("Done")
```

# Continue Statement

```python
n = 10
i = 1
while True:
    if i % 9 != 0:
        print("Inside if")
        i = i+1
        continue
    print("Something")
    print("Something else")
    break        
print("Done")
```
