```python
""" divide the number 146.897 by 12,13,14 and 15 separately, and their quotient and remainder add."""
```

```python
a = divmod(146.897 , 12)
b = divmod(146.897 , 13)
c = divmod(146.897 , 14)
d = divmod(146.897 , 15)
print(a,b,c,d)
```

```python
quotient = a[0]+b[0]+c[0]+d[0]
print(quotient)
```

```python
remainder = a[1]+b[1]+c[1]+d[1]
print(round(remainder))
```

```python
# sum of quotient and remainder
sumofquoandrem = quotient +  remainder
print("The sum of quotient and remainder is:", sumofquoandrem)
```

```python
round(sumofquoandrem, 3)
```

# min & max function 

```python
l = [45,67,43,21,98,-89]
print("Minimum value: ",min(l))
print("Maximum value: ",max(l))
```
